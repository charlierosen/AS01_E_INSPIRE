# Refactored Model Training and DESI Prediction
Configure features and hyperparameters, then train the Random Forest on INSPIRE data before applying it to DESI galaxies.

```python
from pathlib import Path

from src import ModelTrainingConfig, TrainingDatasetConfig, PipelinePaths, RandomForestHyperparameters
from src.training import InspireDesiTrainingPipeline
from src.data import load_desi_catalogue

paths = PipelinePaths()
dataset_config = TrainingDatasetConfig(
    features=['met', 'vdisp', 'tau'],
    restricted=False,
    use_principal_components=False,
)

rf_config = RandomForestHyperparameters(
    n_estimators=50,
    max_depth=8,
    max_features=0.8,
    max_samples=0.7,
    min_samples_leaf=3,
    min_samples_split=5,
    random_state=42,
)

training_config = ModelTrainingConfig(
    paths=paths,
    dataset=dataset_config,
    model=rf_config,
)
pipeline = InspireDesiTrainingPipeline(training_config)
```

```python
artifacts = pipeline.run()
artifacts.desi_predictions.head()
```
